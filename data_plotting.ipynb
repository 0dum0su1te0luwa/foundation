```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_excel(r'C:\Users\USER\Documents\re-integration\datasets\linechart.xlsx')
df.head()
```

```python
plt.figure(figsize=(12,4))
plt.plot(df['Quarter'],df['Fridge'],label='Fridge')
plt.plot(df['Quarter'],df['Dishwasher'],label='Dishwasher')
plt.plot(df['Quarter'],df['Washing Machine'],label='Washing Machine')
plt.xlabel('Quarter')
plt.ylabel('Revenue')
plt.legend()
plt.title('Revenue Chart')
plt.show()
```
